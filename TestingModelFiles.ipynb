```python

import joblib
import pandas as pd

# Load exactly what was saved — nothing else carries over from training
loaded_model = joblib.load("models/rf_model.joblib")          # or "models/lr_pipeline.joblib"
loaded_feature_cols = joblib.load("models/feature_cols.joblib")

print("Model loaded:", type(loaded_model))
print("Expected features:", loaded_feature_cols)
```

```python
# Simulate one new prompt coming in, e.g. from your feature-extraction step.
# Keys must match feature_cols; order doesn't matter here since we pass columns= explicitly.
new_prompt_features = {col: 0 for col in loaded_feature_cols}  # placeholder — replace with real values

X_new = pd.DataFrame([new_prompt_features], columns=loaded_feature_cols)

prediction = loaded_model.predict(X_new)[0]
probability = loaded_model.predict_proba(X_new)[0, 1]

print(f"Prediction: {'ATTACK' if prediction == 1 else 'BENIGN'}")
print(f"Attack probability: {probability:.4f}")
```
